```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df=pd.read_csv('social_media_impact_on_life.csv',encoding='unicode_escape')
df.head()
```

Checking Information Of Data To Analysis

```python
df.info()
```

Checking For Null Values

```python
pd.isnull(df).sum()
```

Filling median at the place of null values

```python
df["Perceived_Stress_Score"] = df["Perceived_Stress_Score"].fillna(
    df["Perceived_Stress_Score"].median()
)

df["Academic_Performance_GPA"] = df["Academic_Performance_GPA"].fillna(
    df["Academic_Performance_GPA"].median()
)
```

```python
pd.isnull(df).sum()
```

```python
print(df["Device_Type"].value_counts())
```

Analysis using BAR graph, Histograph,Pie Chart, Scatter Diagram etc

```python
platform_count = df["Primary_Platform"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(platform_count.index, platform_count.values,
        color=['red', 'yellow', 'orange', 'green', 'purple', 'pink', 'blue'])

plt.title("Most Used Social Media Platforms")
plt.xlabel("Social Media Platform")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)

plt.show()
```

```python
usage = pd.cut(
    df["Daily_Usage_Hours"],
    bins=[0, 2, 4, 6, float("inf")],
    labels=["0–2 Hours", "2–4 Hours", "4–6 Hours", "6+ Hours"]
)
usage_count = usage.value_counts().sort_index()

# Pie chart
plt.figure(figsize=(7,7))

plt.pie(
    usage_count.values,
    labels=usage_count.index,
    autopct="%1.1f%%"
)

plt.title("Daily Social Media Usage")

plt.show()
```

```python
device_count = df["Device_Type"].value_counts()

plt.figure(figsize=(7,5))

plt.bar(device_count.index, device_count.values,
       color=['cyan', 'magenta', 'brown'])

plt.title("Device Used by Students")
plt.xlabel("Device Type")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Perceived_Stress_Score"].dropna(), bins=10,
        color='gray')

plt.title("Distribution of Perceived Stress Score")
plt.xlabel("Stress Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_count = df["Gender"].value_counts()

plt.figure(figsize=(6,6))

plt.figure(figsize=(8,5))

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct="%1.1f%%",
    colors=['orange', 'magenta', 'cyan', 'yellow']
)

plt.title("Gender Distribution")

plt.show()
```

```python
impact_count = df["Overall_Impact"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(impact_count.index, impact_count.values,
       color=['yellow','orange','green'])

plt.title("Overall Impact of Social Media")
plt.xlabel("Overall Impact")
plt.ylabel("Number of Students")

plt.xticks(rotation=45)

plt.show()
```

```python
low_usage = df[df["Daily_Usage_Hours"] <= 4]
high_usage = df[df["Daily_Usage_Hours"] > 4]

low_gpa = low_usage.groupby("Daily_Usage_Hours")["Academic_Performance_GPA"].mean()
high_gpa = high_usage.groupby("Daily_Usage_Hours")["Academic_Performance_GPA"].mean()

plt.figure(figsize=(8,5))

plt.plot(
    low_gpa.index,
    low_gpa.values,
    marker="o",
    label="Low Usage (≤4 Hours)"
)

plt.plot(
    high_gpa.index,
    high_gpa.values,
    marker="o",
    label="High Usage (>4 Hours)"
)

plt.title("Social Media Usage vs Academic Performance")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Average GPA")
plt.legend()
plt.grid(True)

plt.show()
```

```python
avg_sleep = df.groupby("Daily_Usage_Hours")["Sleep_Duration_Hours"].mean()

plt.figure(figsize=(8,5))

plt.plot(
    avg_sleep.index,
    avg_sleep.values,
    color="purple",
    marker="o"
)

plt.title("Daily Social Media Usage vs Average Sleep Duration")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Average Sleep Duration (Hours)")
plt.grid(True)
plt.show()
```
